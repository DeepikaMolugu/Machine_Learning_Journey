```python
import pandas as pd
import requests
```

```python
url =  "https://live-oil-prices.p.rapidapi.com/web-crawling/api/oil-price-charts"

headers = {
    "Content-Type": "application/json",
    "X-RapidAPI-Host":"live-oil-prices.p.rapidapi.com",
    "X-RapidAPI-Key":"d3e2c10462mshfeeeb1c9f2b6cc4p191ab2jsn079d958c7eb2"
}

response = requests.get(url, headers=headers)

df=pd.DataFrame(response.json())
df
```

```python
df.columns=["category","flag_url","commodity_name","current_price","price_change","percent_change","update_info"]
```

```python
df
```

```python
df["percent_change_clean"]=(df["percent_change"].str.extract(r'([+-]?\d+\.\d+)').astype(float))
```

```python
df["delay_minutes"]=(df["update_info"].str.extract(r'(\d+)'))
```

```python
df
```

```python
def category(name):
    if "Crude" in name:
        return "Crude Oil"
    elif "Gas" in name:
        return "Natural Gas"
    elif "Gasoline" in name:
        return "Refined Fuel"
    else:
        return "Other"
df["commodity_type"]=df["commodity_name"].apply(category)
```

```python
df
```

```python
region_map={
    "WTI Crude":"USA",
    "Brent Crude":"Europe",
    "Murban Crude":"UAE",
    "Iran Heavy":"IRAN"
}
df["region"]=df["commodity_name"].map(region_map)
```

```python
df
```

```python
df.drop(columns=['region'],inplace=True)
```

```python
df
```

```python
def volatility(x):
    if abs(x)>5:
        return "High"
    elif abs(x)>2:
        return "Medium"
    return "Low"
df["volatility"]=df["percent_change_clean"].apply(volatility)
```

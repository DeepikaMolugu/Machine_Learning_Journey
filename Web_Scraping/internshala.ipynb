```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url="https://internshala.com/jobs/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "Referer": "https://www.google.com/"
}
response=requests.get(url,headers=headers)
print(response.status_code)
```

```python
soup=BeautifulSoup(response.text,"html.parser")
soup.title
```

```python
container=soup.find_all("div",class_="individual_internship")
len(container)
```

```python
data=[]
for c in container:
    job_title=c.find('h2')
    location=c.find("p",class_="row-1-item locations")
    details=c.find_all("div",class_="row-1-item")
    salary=details[0].text.strip()
    experience=details[1].text.strip()
    key_responsibilities=c.find("div",class_="about_job")
    skills=c.find("div",class_="job_skills")
    data.append({
    "Job Title":job_title.text.strip(),
    "Location":location.text.strip(),
    "Salary":salary,
    "Experience":experience,
    "Responsibilities":key_responsibilities.text.split(),
    "Skills":skills.text.strip()
})

    
    
```

```python
df=pd.DataFrame(data)
df
```

```python
df.to_csv('jobs.csv',index=False)
```

```python
from bs4 import BeautifulSoup
```

```python
def remove_html_tags(text):
    soup=BeautifulSoup(text,"html.parser")
    return soup.get_text()
```

```python
sample=df['review'].iloc[0]
print(remove_html_tags(sample))
```

```python
df['review']=df['review'].apply(remove_html_tags)
```

#### Lowercase Conversion

```python
def convert_to_lowercase(text):
    return text.lower()
```

```python
sample=df['review'].iloc[3]
```

```python
print(convert_to_lowercase(sample))
```

```python
df['review']=df['review'].apply(convert_to_lowercase)
```

#### Punctuation Removal

```python
import string
string.punctuation
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
sample=df['review'].iloc[0]
```

```python
print(remove_punctuation(sample))
```

```python
df['review']=df['review'].apply(remove_punctuation)
```

```python
df['review'].iloc[8]
```

#### Stopwords removal

```python
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
```

```python
print(list(stop_words)[:20])
```

```python
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in stop_words])
```

```python
sample=df['review'].iloc[1]
```

```python
print(remove_stopwords(sample))
```

```python
df['review']=df['review'].apply(remove_stopwords
```

```python
from sklearn.model_selection import train_test_split
```

```python

X_train_text, X_test_text, y_train, y_test = train_test_split(
    df['review'],
    df['sentiment'],
    test_size=0.2,
    random_state=42,
    stratify=df['sentiment']
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()

X_train = cv.fit_transform(X_train_text)
X_test = cv.transform(X_test_text)
```

```python
from sklearn.naive_bayes import MultinomialNB

mnb = MultinomialNB()

mnb.fit(X_train, y_train)
```

```python
y_pred = mnb.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

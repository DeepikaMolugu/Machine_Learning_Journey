```python
def run_score(batsman_name):
    mask=df2['batsman']==batsman_name
    new_mask=df2[mask]
    return new_mask.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(1)
    
```

```python
runs_conceded=df2.groupby('bowler')[['batsman_runs','wide_runs','noball_runs']].sum().sum(axis=1)
runs_conceded
```

#### Economy Rate=Runs Conceded/Overs Bowled

```python
legal_balls=df2[(df2['wide_runs']==0) & (df2['noball_runs']==0)]
balls=legal_balls.groupby('bowler').size()
overs_bowled=balls/6
overs_bowled
```

```python
economy_rate=runs_conceded/overs_bowled
```

```python
economy_rate
```

```python
economy_rate[(economy_rate < 7)&(overs_bowled>=20)].sort_values(ascending=False)
```

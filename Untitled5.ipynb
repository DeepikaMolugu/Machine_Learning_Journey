```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('ev_market_2026.csv')
df
```

```python
df=df.iloc[:,4:11]
```

```python
df
```

```python
df.head()
```

```python
df.columns
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('price_usd',axis=1),df['price_usd'],test_size=0.3,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
X_train
```

```python
X_train_scaled
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
ax1.scatter(X_train['battery_capacity_kwh'],X_train['acceleration_0_60_mph'])
ax1.set_title("Before Scaling")
ax2.scatter(X_train_scaled['battery_capacity_kwh'],X_train_scaled['acceleration_0_60_mph'],color='red')
ax2.set_title("After Scaling")
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Before Scaling')
sns.kdeplot(X_train['battery_capacity_kwh'], ax=ax1)
sns.kdeplot(X_train['acceleration_0_60_mph'], ax=ax1)

# after scaling
ax2.set_title('After Standard Scaling')
sns.kdeplot(X_train_scaled['battery_capacity_kwh'], ax=ax2)
sns.kdeplot(X_train_scaled['acceleration_0_60_mph'], ax=ax2)
plt.show()
```

```python

fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Acceleration  Before Scaling')
sns.kdeplot(X_train['acceleration_0_60_mph'], ax=ax1)

# after scaling
ax2.set_title('Acceleration After Standard Scaling')
sns.kdeplot(X_train_scaled['acceleration_0_60_mph'], ax=ax2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Battery Capacity Distribution Before Scaling')
sns.kdeplot(X_train['battery_capacity_kwh'], ax=ax1)

# after scaling
ax2.set_title('Battery Capacity Distribution Standard Scaling')
sns.kdeplot(X_train_scaled['battery_capacity_kwh'], ax=ax2)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
lr_scaled = LinearRegression()
lr_scaled.fit(X_train_scaled,y_train)
```

```python
y_pred=lr.predict(X_test)
y_pred_scaled=lr_scaled.predict(X_test_scaled)
```

```python
from sklearn.metrics import r2_score

print("Actual:", r2_score(y_test, y_pred))
print("Scaled:", r2_score(y_test, y_pred_scaled))
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(np.mean(arr))   # Average
print(np.median(arr)) # Middle value
print(np.std(arr))    # Standard deviation
print(np.var(arr))    # Variance
```

```python
arr = np.array([1, 2, 10, 3, 4])

print(np.mean(arr))   # (1+2+10+3+4)/5 = 4
print(np.median(arr)) # sorted → [1,2,3,4,10] → 3
```

```python
arr = np.array([1, 2, 3, 4, 5])

print(np.var(arr))  # Variance
print(np.std(arr))  # Standard deviation
```

```python
arr = np.array([10, 20, 5, 40])

print(np.max(arr))  # 40
print(np.min(arr))  # 5
print(np.ptp(arr))  # range = max - min = 35
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(np.percentile(arr, 25))  # 25th percentile
print(np.percentile(arr, 50))  # median
print(np.percentile(arr, 90))  # 90th percentile
```
